### Part 1: Creative text generation with a Hugging Face model.

*** This script uses the free Hugging Face Transformers library with the
distilgpt2 text-generation model. It writes the generated creative output to
generated_creative_output.txt as required by the assignment.
***

```python
from pathlib import Path

from transformers import pipeline, set_seed

# Creative prompt required by the assignment.
PROMPT = (
    "Write a short imaginative story about a small robot learning to dream "
    "for the first time in a quiet city after midnight."
)

OUTPUT_FILE = Path("generated_creative_output.txt")

# distilgpt2 is a small open-source Hugging Face text-generation model.
# Key parameters:
# - max_new_tokens controls the length of the generated continuation.
# - temperature and top_p make the output more creative while still focused.
# - no_repeat_ngram_size reduces repeated phrases.
generator = pipeline("text-generation", model="distilgpt2")
set_seed(42)

result = generator(
    PROMPT,
    max_new_tokens=180,
    temperature=0.9,
    top_p=0.95,
    do_sample=True,
    no_repeat_ngram_size=2,
    pad_token_id=generator.tokenizer.eos_token_id,
)

generated_text = result[0]["generated_text"].strip()
OUTPUT_FILE.write_text(generated_text + "\n", encoding="utf-8")

print(f"Creative text saved to {OUTPUT_FILE}")
```
